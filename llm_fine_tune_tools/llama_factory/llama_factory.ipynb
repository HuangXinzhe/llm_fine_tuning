# llama_factory
详细介绍见：https://github.com/hiyouga/LLaMA-Factory/blob/main/README_zh.md

## 1. 登录Huggingface账号
```
pip install --upgrade huggingface_hub
huggingface-cli login
```

## 2. 安装依赖
```
git clone https://github.com/hiyouga/LLaMA-Factory.git
conda create -n llama_factory python=3.10
conda activate llama_factory
cd LLaMA-Factory
pip install -e .[metrics]
```

## 3. 模型训练
### 利用 LLaMA Board 可视化界面训练（由 Gradio 驱动）
export CUDA_VISIBLE_DEVICES=0 # Windows 使用 `set CUDA_VISIBLE_DEVICES=0`
export GRADIO_SERVER_PORT=7860 # Windows 使用 `set GRADIO_SERVER_PORT=7860`
python src/train_web.py # 或 python -m llmtuner.webui.interface

### 利用命令行接口训练
使用方法请参考 examples/README_zh.md。

您可以执行 python src/train_bash.py -h 来查看参数文档。

### 2. 克隆llama_factory仓库，并安装依赖
```
git clone https://github.com/hiyouga/LLaMA-Factory.git

# 进入工作目录
cd LLaMA-Factory

# 安装依赖
pip install -e .[metrics]
```

### 3. 设置参数
```
# 设置CUDA变量
export CUDA_VISIBLE_DEVICES=0

# 设置默认端口7860
export GRADIO_SERVER_PORT=7860

# 设置模型下载源，魔塔
expert USE_MODELSCOPE_HUB=1

# 模型下载的路径\root\.cache\modelscope\hub

# 启动微调UI
python src/train_web.py
```

- 进入http://localhost:7860/
- 选择模型
- 选择数据集
- 设置超参数
- 开始微调
- 可以在chat选项卡，选择加载方式，直接部署出来，测试效果！
- 如果效果达到了，可以在下方这个选项卡，选择量化方式，直接合并导出模型！
